```python
def convert_units(conversion_type, value):
  '''
  This function performs conversion based on the selected type.

  Parameters:
  conversion_type (int): This is the specific conversion formula to apply.
  value (value): This is the numeric value to be converted.

  Returns:
  float: The converted value
  '''
  # Length conversions

  # m to ft
  if conversion_type == 1: return value *3.28084
  #ft to m
  if conversion_type == 2: return value / 3.28084

  # Weight conversions

  # kg to lbs
  if conversion_type == 3: return value * 2.20462
  # lbs to kg
  if conversion_type == 4: return value / 2.20462

  # Volume conversions

  # L to gal
  if conversion_type == 5: return value * 0.264172
  # gal to L
  if conversion_type == 6: return value / 0.264172

def main():
  print("---  Unit Converter ---")
  print("1: Meters to Feet    \n2: Feet to Meters")
  print("3: KG to Pounds      \n4: Pounds to KG")
  print("5: Liters to Gallons \n6: Gallons to Liters")

  try:
    choice = int(input("\nSelect any conversion (1-6): "))

    if choice not in range(1,7):
      print("Error. Choose between the given range.")
      return

    val = float(input("Enter any value to convert: "))

    result = convert_units(choice, val)
    print(f"Result: {result:.2f}")

  except ValueError:
    print("Invalid input! Please enter numbers only.")


main()

```

```python
def calculate_sum(numbers):
  '''
  Calculates the total sum of a list of numbers.

  Parameters:
  numbers (list): A list of numeric values (int or float) to be summed.

  Returns:
  float/int: The total sum of the numbers.

  '''
  return sum(numbers)

def calculate_avg(numbers):
  '''
  Calculates the average of a list of numbers.

  Parameters:
  numbers (list): A list of numeric values to be averaged.

  Returns:
  float: The arithmetic mean of the numbers.

  '''
  return sum(numbers) / len(numbers)

def calculate_max(numbers):
  '''
  Finds the largest number in the list.

  Parameters:
  numbers (list): A list of numeric values to search.

  Returns:
  float/int: The maximum value found in the list.

  '''
  return max(numbers)

def calculate_min(numbers):
  '''
  Finds the smallest number in the list.

  Parameters:
  numbers (list): A list of numeric values to search.

  Returns:
  float/int: The minimum value found in the list.

  '''
  return min(numbers)

print("---  Math Operations ---")
print("Options: \n1: Sum \n2: Average \n3: Maximum \n4: Minimum")

try:
    choice = int(input("\nSelect an operation (1-4): "))

    if choice not in [1, 2, 3, 4]:
        print("Invalid choice. Please pick 1, 2, 3, or 4.")

    else:
        raw_input = input("Enter numbers separated by spaces (e.g., 10 20 30): ")
        num_list = [float(x) for x in raw_input.split()]

        if not num_list:
            print("The list is empty! Please enter some numbers.")
        else:
            # Perform the selected operation
            if choice == 1:
                print(f"Sum: {calculate_sum(num_list)}")
            elif choice == 2:
                print(f"Average: {calculate_avg(num_list)}")
            elif choice == 3:
                print(f"Maximum: {calculate_max(num_list)}")
            elif choice == 4:
                print(f"Minimum: {calculate_min(num_list)}")

except ValueError:
    print("Error: Please enter valid numbers only.")
```

## 4.2 Exercise on List Manipulation

```python
#This is the list being tested for every list manipulation exercise
test_list = [1, 2, 3, 4, 5, 6]
```

### 1. Extract Every Other Element:

```python
def extract_every_other(lst):
    """Returns a new list containing every other element starting from the first."""
    return lst[::2]
print("Every other:", extract_every_other(test_list))
```

### 2. Slice a Sublist:

```python
def get_sublist(lst, start, end):
    """Returns the sublist from start to end inclusive using slicing."""
    return lst[start : end + 1]

print("Sublist:", get_sublist(test_list, 2, 4))
```

### 3. Reverse a List Using Slicing

```python
def reverse_list(lst):
    """Reverses a list using slicing."""
    return lst[::-1]

print("Reversed List is:", reverse_list(test_list))
```

### 4. Remove the First and Last Elements

```python
def remove_first_last(lst):
    """Removes the first and last elements using slicing."""
    return lst[1:-1]

print("After removing first and last elements:", remove_first_last(test_list))
```

### 5. Get the First n Elements

```python
def get_first_n(lst, n):
    """Extracts the first n elements using slicing."""
    return lst[:n]

print("After getting the first 'n' elements:", get_first_n(test_list,3))
```

### 5. Get the Last n Elements

```python
def get_last_n(lst, n):
    """Extracts the last n elements using slicing."""
    return lst[-n:]

print("After getting the last 'n' elements:", get_last_n(test_list,3))
```

### 7. Extract Elements in Reverse Order:

```python
def reverse_skip(lst):
    """Returns every second element starting from the second-to-last, moving backward."""
    return lst[-2::-2]

print("After reverse skip:", reverse_skip(test_list))
```

## 4.3 Exercise on Nested List:

```python
nest_list = [[1, 2, 2, 3], [4, 5, 6], [7, 8, 9]]
```

### 1. Flatten a Nested List:

```python
def flatten(lst):
  """Flattens a 2D nested list into a single list."""
  flattened = []
  for sublist in lst:
    for item in sublist:
      flattened.append(item)
  return flattened


print("The flattened list is:",flatten(nest_list))
```

### 2. Accessing Nested List Elements:

```python
def access_nested_element(lst, indices):
    """Extracts an element from a nested list given a list of indices."""
    result = lst
    for index in indices:
        result = result[index]
    return result

print("The extracted element is:",access_nested_element(nest_list, [1,2]))
```

### 3. Sum of All Elements in a Nested List:

```python
def sum_nested(lst):
    """Calculates the sum of all numbers in a nested list of any depth."""
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total

print("The sum of all the elements are:",sum_nested(nest_list))
```

### 4. Remove Specific Element from a Nested List:

```python
def remove_element(lst, elem):
    """Removes all occurrences of elememts from the nested list."""
    result = []
    for item in lst:
        if isinstance(item, list):
            result.append(remove_element(item, elem))
        elif item != elem:
            result.append(item)
    return result

print("The list after removing a specific element is:",remove_element(nest_list,6))
```

### 5. Find the Maximum Element in a Nested List:

```python
def find_max(lst):
    """Finds the maximum element in a nested list of any depth."""
    current_max = float('-inf')
    for item in lst:
        if isinstance(item, list):
            res = find_max(item)
        else:
            res = item
        if res > current_max:
            current_max = res
    return current_max

print("The maximum number from the list is:",find_max(nest_list))
```

### 6. Count Occurrences of an Element in a Nested List:

```python
def count_occurrences(lst, elem):
    """Counts how many times elem appears in a nested list."""
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        elif item == elem:
            count += 1
    return count

print("The element appears following times:",count_occurrences(nest_list,2))
```

### 7. Flatten a List of Lists of Lists:

```python
def deep_flatten(lst):
    """Flattens a list of lists of lists... into a single list."""
    flattened = []
    for item in lst:
        if isinstance(item, list):
            flattened.extend(deep_flatten(item))
        else:
            flattened.append(item)
    return flattened

print("The flattened list of lists is:",deep_flatten(nest_list))
```

### 8. Nested List Average:

```python
def average_nested(lst):
    """Calculates the average of all elements in a nested list."""
    flat_list = deep_flatten(lst)
    if not flat_list:
        return 0
    return sum(flat_list) / len(flat_list)

print("The average of the list is:",average_nested(nest_list))
```

```python
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

# 1. Prove A * A^-1 = I
A_inv = np.linalg.inv(A)
identity_check = np.dot(A, A_inv)
print("1. A * inv(A) (Identity Matrix):\n", np.round(identity_check))

# 2. Prove AB != BA
AB = np.dot(A, B)
BA = np.dot(B, A)
print(f"\n2. Prove AB != BA:")
print("   AB:\n", AB)
print("   BA:\n", BA)
print("   Are they equal?", np.array_equal(AB, BA))

# 3. Prove (AB)^T = B^T * A^T
lhs = (AB).T
rhs = np.dot(B.T, A.T)
print(f"\n3. Transpose Proof (AB)^T == B^T @ A^T: {np.allclose(lhs, rhs)}")
```

## 10.2 Experiment: How Fast is Numpy?

### 1. Element-wise Addition:

```python
import time

# Set the size for experiments 1, 2, and 3
size = 1000000

# Prepare data for Python Lists
list1 = list(range(size))
list2 = list(range(size))

# Prepare data for NumPy Arrays
arr1 = np.arange(size)
arr2 = np.arange(size)

print("1. Element-wise Addition")

# Python List Addition (requires a loop or list comprehension)
start = time.time()
result_list = [list1[i] + list2[i] for i in range(size)]
print(f"Python List Time: {time.time() - start:.5f} seconds")

# NumPy Array Addition
start = time.time()
result_arr = arr1 + arr2
print(f"NumPy Array Time: {time.time() - start:.5f} seconds")
```

### 4. Matrix Multiplication

```python
import time
import random

n = 1000

# create matrices
A = [[random.random() for j in range(n)] for i in range(n)]
B = [[random.random() for j in range(n)] for i in range(n)]

# result matrix
C = [[0 for j in range(n)] for i in range(n)]

start = time.time()

for i in range(n):
    for j in range(n):
        for k in range(n):
            C[i][j] += A[i][k] * B[k][j]

end = time.time()

print("Time using Python lists:", end - start, "seconds")
```
